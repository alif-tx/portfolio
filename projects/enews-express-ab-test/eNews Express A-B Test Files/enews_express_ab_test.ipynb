# **STRATEGIC SUMMARY & PROJECT GENESIS**

## **Context**  

The digital age has ushered in a new era for news consumption: e-news portals. These platforms offer an unparalleled opportunity to stay abreast of global events, delivering information swiftly and efficiently. News is gathered from online databases, processed through sophisticated software, and then transmitted to users, offering advantages such as rapid access and the integration of multimedia elements like audio, graphics, video, and other features often absent or underdeveloped in traditional print media.

**𝕖-𝙽𝙴𝚆𝚂 𝙴𝚇𝙿𝚁𝙴𝚂𝚂**, a prominent online news portal, is eager to expand its subscriber base. The company observes that visitors engage with its website in various ways, and by analyzing these interactions, **𝕖-𝙽𝙴𝚆𝚂 𝙴𝚇𝙿𝚁𝙴𝚂𝚂**  aims to better understand user interests and enhance engagement. Leadership at **𝕖-𝙽𝙴𝚆𝚂 𝙴𝚇𝙿𝚁𝙴𝚂𝚂** suspects a decline in new monthly subscribers over the past year, attributing it to a current webpage design that may not sufficiently engage users to encourage subscription.

To address this, **𝕖-𝙽𝙴𝚆𝚂 𝙴𝚇𝙿𝚁𝙴𝚂𝚂** conducted an A/B test on the two designs, with the **subscription conversion rate** as the primary scorecard. The methodology is built to separate a real effect of the redesign from random noise in the sample.

## **Objective**

The design team at **𝕖-𝙽𝙴𝚆𝚂 𝙴𝚇𝙿𝚁𝙴𝚂𝚂** has developed a new landing page, featuring a refreshed layout and more pertinent content than its predecessor. To evaluate the effectiveness of this new design in attracting subscribers, the Data Science team conducted an experiment. They randomly selected 100 users, dividing them equally into two groups: a **control group**, which was presented with the existing landing page, and a **treatment group**, which encountered the new landing page. Data on user interactions with both versions of the page were systematically collected.

As a data scientist at **𝕖-𝙽𝙴𝚆𝚂 𝙴𝚇𝙿𝚁𝙴𝚂𝚂**, your task is to explore this dataset and perform a statistical analysis (at a **`significance level of 5%`**) to ascertain whether the new landing page is more effective at converting new subscribers. This will
involve answering the following key questions:

* **Objective 1: Engagement Lift** — Do users spend more time on the new landing page than the existing landing page?
* **Objective 2: Subscriber Growth** — Do users convert at a higher rate on the new landing page than the existing landing page?
* **Objective 3: Linguistic Independence** — Does the converted status depend on the preferred language?
* **Objective 4: Cross-Linguistic Engagement** — Is the time spent on the new page the same for different language users?

```python
# ── Bivariate Num-Num conversion time spent on page log ────────────────────────────────
hout(hban("TIME SPENT × CONVERSION STATUS — ENGAGEMENT CORRELATION",
          "Deciles Heatmap and Cumulative Conversion Line"))

# ── Encode conversion as numeric and rank time-on-page for decile binning ──────────────
df['converted_binary'] = df['converted'].map({'yes': 1, 'no': 0})
df['time_rank']     = df['time_spent_on_the_page'].rank(method='first').astype(int)
df['rank_decile']   = pd.cut(df['time_rank'], bins=10,
                              labels=['D1','D2','D3','D4','D5','D6','D7','D8','D9','D10'])

# ── Build a 1×2 figure: decile heatmap on left, ranked bar + cumulative line on right ──
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# ── Left panel: heatmap of conversion counts across time-rank deciles ──────────────────
ct_heat = pd.crosstab(df['rank_decile'], df['converted'],
                       rownames=['Rank Decile'], colnames=['Converted'])
ct_heat = ct_heat[['yes', 'no']]
sns.heatmap(ct_heat, annot=True, fmt='d', cmap='RdPu', vmin=0, vmax=10,
            linecolor=ACCENT, linewidths=0.5,
            annot_kws={'fontweight': 'bold'}, ax=axes[0])
axes[0].set_title('Conversion Decision by Time Rank Decile', fontsize=14,
                   fontweight='bold', pad=20)
axes[0].title.set_position([0.45, 1.05])
axes[0].set_xlabel('Converted', fontsize=11, fontweight='bold', labelpad=12)
axes[0].set_ylabel('Rank Decile  (D1 = least time → D10 = most time)',
                    fontsize=10, fontweight='bold', labelpad=12)
axes[0].set_xticklabels(['Yes', 'No'], fontweight='bold', rotation=0)
axes[0].set_yticklabels(axes[0].get_yticklabels(), fontweight='bold', rotation=0)

# ── Right panel: ranked bar chart with cumulative conversion rate overlay ──────────────
df_ranked  = df.sort_values('time_rank').reset_index(drop=True)
bar_colors = [HPRIMARY if c == 1 else '#223444' for c in df_ranked['converted_binary']]
axes[1].bar(df_ranked['time_rank'], df_ranked['converted_binary'],
            color=bar_colors, edgecolor='none', width=1.0)

# ── Cumulative conversion line on secondary y-axis ─────────────────────────────────────
cum_conv = df_ranked['converted_binary'].expanding().mean() * 100
ax2 = axes[1].twinx()
ax2.plot(df_ranked['time_rank'], cum_conv, color=ACCENT, linewidth=2.5,
         label='Cumulative Conversion Rate (%)')
ax2.set_ylabel('Cumulative Conversion Rate (%)', fontsize=10,
               fontweight='bold', labelpad=12, color=ACCENT)
ax2.tick_params(axis='y', labelcolor=ACCENT)
ax2.set_ylim(0, 100)

# ── Right panel axis labels and styling ────────────────────────────────────────────────
axes[1].set_title('Conversion Decision by Time Rank', fontsize=14,
                   fontweight='bold', pad=20)
axes[1].title.set_position([0.45, 1.05])
axes[1].set_xlabel('Time Rank  (1 = least → 100 = most)', fontsize=11,
                    fontweight='bold', labelpad=12)
axes[1].set_ylabel('Converted  (1 = Yes, 0 = No)', fontsize=11,
                    fontweight='bold', labelpad=12)
axes[1].set_yticks([0, 1])
axes[1].set_yticklabels(['No (0)', 'Yes (1)'], fontweight='bold')
ax2.legend(loc='lower right', framealpha=0.9)
sns.despine(ax=axes[1])
plt.tight_layout()
plt.show()

# ── Spearman rank correlation summary with mean rank by conversion status ──────────────
rho, p_rho = spearmanr(df['time_rank'], df['converted_binary'])
mean_yes = df[df.converted_binary == 1].time_rank.mean()
mean_no  = df[df.converted_binary == 0].time_rank.mean()
hout(
    f"Time spent on the page and conversion show a Spearman rank correlation of "
    f"{hval(f'{rho:.4f}')} with a <i>p</i>-value of {hval(f'{p_rho:.4f}')}. Users who converted "
    f"sat at an average time rank of {hval(f'{mean_yes:.1f}')} compared to "
    f"{hval(f'{mean_no:.1f}')} for those who did not, suggesting that longer engagement "
    f"may be associated with the decision to subscribe."
)
```

## **i — Define the Null and Alternate Hypotheses**

* **H₀:** μ_new ≤ μ_old — Mean time on the new page is less than or equal to that of the old page.
* **Hₐ:** μ_new > μ_old — Mean time on the new page is **greater than** that of the old page.
* **Direction: One-tailed (right-tailed)**

## **ii — Select the Appropriate Test**

* **Test Name:** **Welch's Two-Sample t-test**
* **Rationale:** Two independent groups, continuous variable, unknown and potentially unequal population variances. Sample standard deviations will be compared to justify `equal_var=False`.

## **iii — Decide the Significance Level**

* $\alpha = 0.05$ — A baseline 5% maximum risk of committing a Type I error (false positive verdict) is accepted.  

## **iv — Collect and Prepare Data**

```python
# ── Stripe header for executive summary output ─────────────────────────────────────────
hout(hban("PARAMETRIC ASSUMPTION VERIFICATION",
          "Sample Statistics and Normality / Variance Diagnostics — Objective 1"))

# ── Per-group data and t-test assumption checks ────────────────────────────────────────
# Check normality (Shapiro-Wilk) and equal variance (Levene) to choose pooled vs Welch's t-test
time_new = df[df['landing_page'] == 'new']['time_spent_on_the_page']
time_old = df[df['landing_page'] == 'old']['time_spent_on_the_page']

summary_q1 = (df.groupby('landing_page')['time_spent_on_the_page']
                .agg(['count','mean','std'])
                .reindex(['new','old'])
                .round(4))

w_new, p_new     = shapiro(time_new)
w_old, p_old     = shapiro(time_old)
lev_q1, p_lev_q1 = levene(time_new, time_old)

# ── Dynamic verdict helpers driven by project-wide alpha ───────────────────────────────
def norm_verdict(p):
    return f'{hval("[X]")} — Normal' if p > alpha else f'{hval("[ ]")} — Non-normal'

var_q1          = f'{hval("[X]")} — Equal variances' if p_lev_q1 > alpha else f'{hval("[ ]")} — Unequal variances'
independence_ok = True
indep_verdict   = f'{hval("[X]")} — Accepted' if independence_ok else f'{hval("[ ]")} — Not Accepted'
_grand          = df['time_spent_on_the_page'].mean()
_mean_diff      = time_new.mean() - time_old.mean()

# ── Consolidated verdict: Welch correction kicks in automatically if variances differ ──
both_normal_q1 = (p_new > alpha) and (p_old > alpha)
if both_normal_q1 and independence_ok:
    if p_lev_q1 > alpha:
        verdict_marker_q1    = f'{hval("[X]")} — Parametric assumptions met'
        verdict_procedure_q1 = "Standard two-sample t-test is the appropriate inferential procedure, applied with equal_var=True"
    else:
        verdict_marker_q1    = f'{hval("[X]")} — Parametric assumptions met'
        verdict_procedure_q1 = "Welch's t-test is the appropriate inferential procedure, applied with equal_var=False to correct for unequal variances"
else:
    verdict_marker_q1    = f'{hval("[ ]")} — Normality assumption failed'
    verdict_procedure_q1 = "consider Mann-Whitney U as a non-parametric alternative"

# ── Label-to-marker column so every [X] vertically stacks at the same position ─────────
L_grand  = " • Grand mean (all users):"
L_diff   = " • Observed mean difference (new − old):"
L_new    = " • Shapiro-Wilk (New page):"
L_old    = " • Shapiro-Wilk (Old page):"
L_lev    = " Levene's Test for Homogeneity of Variance"
L_indep  = " Independence Requirement Satisfied (Random Assignment)"
L_verd   = " Final assumption verdict"

# ── Parametric Assumption Verification logs (monospace diagnostics) ────────────────────
hout(
    f"{OSTYLE1}"
        f"{L_grand.ljust(MARKER_COL)}{hval(f'{_grand:.4f} min')}\n"
        f"{L_diff.ljust(MARKER_COL)}{hval(f'{_mean_diff:.4f} min')}\n"
        f"{L_new.ljust(MARKER_COL)}{norm_verdict(p_new)}\n"
        f"   W={hval(f'{w_new:.4f}')}, p={hval(f'{p_new:.4f}')}\n"
        f"{L_old.ljust(MARKER_COL)}{norm_verdict(p_old)}\n"
        f"   W={hval(f'{w_old:.4f}')}, p={hval(f'{p_old:.4f}')}"
        f"<hr style='max-width: 720px; margin: 8px 0; border: 0; border-top: 1px solid #ddd;'>"
        f"<b>{L_lev.ljust(MARKER_COL)}{var_q1}</b>\n"
        f"<b>{L_indep.ljust(MARKER_COL)}{indep_verdict}</b>"
        f"<hr style='max-width: 720px; margin: 8px 0; border: 0; border-top: 1px solid #ddd;'>"
        f"<b>{L_verd.ljust(MARKER_COL)}{verdict_marker_q1}</b>\n\n"
        f"<b> {verdict_procedure_q1}</b>"
    f"</pre>"
)
display(summary_q1)
```

## **v — Calculate the *p*-value**

```python
# ── Stripe header for hypothesis test results ──────────────────────────────────────────
hout(hban("HYPOTHESIS TEST RESULTS",
          "Welch's Two-Sample t-test (One-Tailed) — Objective 1"))

# ── Welch's Two-Sample t-test execution: one-tailed p-value + two-sided 95% CI ───────────
#    Welch's chosen for unequal variances (Levene p < α); alternative='greater' for right-tailed H_a
result_q1 = ttest_ind(time_new, time_old, equal_var=False, alternative='greater')
result_q1   = ttest_ind(time_new, time_old, equal_var=False, alternative='greater')
t_stat_q1   = result_q1.statistic
p_val_q1    = result_q1.pvalue

# ── Two-sided 95% CI for the mean difference (conventional A/B reporting convention) ───
ci_q1       = ttest_ind(time_new, time_old, equal_var=False).confidence_interval(confidence_level=0.95)
ci_low_q1   = ci_q1.low
ci_high_q1  = ci_q1.high

hout(
    f"Executing a Welch's Two-Sample t-test on the continuous engagement data yields a "
    f"formal test statistic of {hval(f'{t_stat_q1:.4f}')} and a corresponding <i>p</i>-value of "
    f"{hval(f'{p_val_q1:.4g}')}. The two-sided 95% confidence interval for the population mean "
    f"difference (new − old) is {hval(f'{ci_low_q1:.4f}')} to {hval(f'{ci_high_q1:.4f}')} minutes."
)
```

## **vi — Compare the *p*-value with $α$**

```python
# ── Stripe header for significance evaluation ──────────────────────────────────────────
hout(hban("SIGNIFICANCE EVALUATION","Alpha Threshold Comparison — Objective 1"))

# ── Compare p-value to significance level ──────────────────────────────────────────────
# Decision rule: reject H₀ if p_val_q1 < α
decision_q1 = 'REJECT the Null Hypothesis' if p_val_q1 < alpha else 'FAIL TO REJECT the Null Hypothesis'

hout(
    f"Comparing the resulting <i>p</i>-value of {hval(f'{p_val_q1:.4g}')} against the established significance level of "
    f"α = {hval(alpha)} results in a mathematical decision to {hval(decision_q1)}."
)
```

---

## **i — Define the Null and Alternate Hypotheses**
* **H₀:** $p_{\text{new}} \le p_{\text{old}}$ — The conversion rate for the new landing page is less than or equal to the old landing page rate.
* **Hₐ:** $p_{\text{new}} > p_{\text{old}}$ — The conversion rate for the new landing page is significantly **greater than** the old landing page rate.
* **Direction:** One-tailed (Right-tailed verification)
<br>

## **ii — Select the Appropriate Test**
* **Test Name:** **Two-Proportion Z-test**
* **Rationale:** The evaluation compares independent binomial population vectors measuring a binary categorical outcome (`converted: yes/no`) across two exclusive treatment groups. Executed using `alternative='larger'` with treatment successes passed as the primary array.
<br>

## **iii — Decide the Significance Level**
* $\alpha = 0.05$ (a baseline 5% maximum risk of committing a Type I error / false positive verdict is accepted).  

## **iv — Collect and Prepare Data**

```python
# ── Stripe header for executive summary output ─────────────────────────────────────────
hout(hban("PARAMETRIC ASSUMPTION VERIFICATION","Sample Statistics and Binomial Diagnostics — Objective 2"))

# ── Conversion counts and normal approximation check ───────────────────────────────────
# Verify normal-approx conditions (n·p ≥ 10 and n·(1−p) ≥ 10 per group) for two-proportion z-test
n_treat = df[df["group"] == "treatment"].shape[0]
n_ctrl = df[df["group"] == "control"].shape[0]

# ── Extraction of observed group successes ─────────────────────────────────────────────
conv_new = (df[df["group"] == "treatment"]["converted"] == "yes").sum()
conv_old = (df[df["group"] == "control"]["converted"] == "yes").sum()

# ── Calculation of sample proportion vectors ───────────────────────────────────────────
p_new_obs = conv_new / n_treat
p_old_obs = conv_old / n_ctrl

# ── Binomial criteria validation sequence ──────────────────────────────────────────────
rows = []
q2iv_passed = True  # Guardrail tracking variable

for label, n, p in [
    ("Treatment (new)", n_treat, p_new_obs),
    ("  Control (old)", n_ctrl, p_old_obs),
]:
    np_ = n * p
    nq_ = n * (1 - p)

    if np_ < 10 or nq_ < 10:
        q2iv_passed = False

    ok2 = "[X]" if nq_ >= 10 else "[ ]"
    ok3 = " — Accepted" if nq_ >= 10 else " — Not Accepted"

# ──  Group Diagnostic Summary Details  ─────────────────────────────────────────────────
    rows.append(
        f"<pre><div style='display: flex; justify-content: space-between; max-width: 620px; margin-bottom: 4px;'>"
        f"  <span>&bull; {label}: n &times; p = {hval(f'{np_:.1f}')}; &emsp; n &times; (1 &minus; p) = {hval(f'{nq_:.1f}')}</span>"
        f"  <span>{hval(ok2)}<strong>{ok3}</strong></span>"
        f"</div></pre>"
    )

# ── Setup substitution parameters based on the guardrail state ─────────────────────────
q2iv_summary_marker = "[X]" if q2iv_passed else "[ ]"
q2iv_summary_text = (
    " — Accepted" if q2iv_passed else " — Requirements Not Satisfied"
)

# ── Render structures for italicized proportion variables ──────────────────────────────
p_new_latex = f'<i style="font-family: serif; font-size: 15px;">p</i><sub>new</sub>'
p_old_latex = f'<i style="font-family: serif; font-size: 15px;">p</i><sub>old</sub>'

# ── Compilation of unified document markup ─────────────────────────────────────────────
hout(f"""
<b>Sample Statistics</b><br>
Treatment: {hval(f'{conv_new}/{n_treat}')} converted &rarr; {p_new_latex} = {hval(f'{p_new_obs:.4f}')} ({hval(f'{p_new_obs*100:.1f}%')})<br>
Control: {hval(f'{conv_old}/{n_ctrl}')} converted &rarr; {p_old_latex} = {hval(f'{p_old_obs:.4f}')} ({hval(f'{p_old_obs*100:.1f}%')})<br>
Observed difference: {hval(f'{(p_new_obs - p_old_obs)*100:.1f} pp')}<br><br>

<b>Normal Approximation Check (n &times; p &ge; 10, n &times; (1 &minus; p) &ge; 10):</b><br>
{"".join(rows)}
<hr>
<pre><div style="display: flex; justify-content: space-between; font-weight: bold; max-width: 620px; margin-bottom: 4px;">
    <span>Both cohorts satisfy normal approximation requirements</span>
    <span>{hval(q2iv_summary_marker)}{q2iv_summary_text}</span>
</div></pre>
""")
```

## **v — Calculate the Test Statistic and *p*-value**

```python
# ── Stripe header for executive summary output ─────────────────────────────────────────
hout(hban("PARAMETRIC PROPORTION ESTIMATION PIPELINE","Right-Tailed Two-Proportion Z-Test Execution — Objective 2"))

# ── Execution of the Two-Proportion Z-test ─────────────────────────────────────────────
counts = np.array([conv_new, conv_old])
nobs = np.array([n_treat, n_ctrl])

# ── Compute test statistic and right-tailed p-value ────────────────────────────────────
# Two-proportion z-test (right-tailed); normal-approx conditions verified above
z_stat_q2, p_val_q2 = proportions_ztest(counts, nobs, alternative="larger")

# ── Compilation of executive summary markup ────────────────────────────────────────────
hout(f"""
<pre><div style="display: flex; justify-content: space-between; font-weight: bold; max-width: 620px; margin-bottom: 4px;">
    <span>&bull; Calculated Z-Statistic:</span>
    <span>{hval(f'{z_stat_q2:.4f}')}</span>
</div></pre>

<pre><div style="display: flex; justify-content: space-between; font-weight: bold; max-width: 620px; margin-bottom: 4px;">
    <span>&bull; Calculated Asymptotic <i>P</i>-Value:</span>
    <span>{hval(f'{p_val_q2:.6f}')}</span>
</div></pre>
""")
```

## **vi — Compare the *p*-value with $α$**

```python
# ── Stripe header for critical value evaluation ────────────────────────────────────────
hout(hban("CRITICAL VALUE CRITERIA EVALUATION",
          f"Significance Threshold Verification (α = {alpha}) — Objective 2"))

# ── Alpha criteria threshold evaluation ────────────────────────────────────────────────
# Decision rule: reject H₀ if p_val_q2 < α
q2vi_passed = p_val_q2 < alpha
q2vi_summary_marker = "[X]" if q2vi_passed else "[ ]"
q2vi_summary_text = " — Reject H₀" if q2vi_passed else " — Fail to Reject H₀"

# ── Render result ──────────────────────────────────────────────────────────────────────
hout(f"""
<pre><div style="display: flex; justify-content: space-between; font-weight: bold; max-width: 620px; margin-bottom: 4px;">
    <span>&bull; Alpha Threshold Check (<i>p</i>-value &lt; &alpha;):</span>
    <span>{hval(q2vi_summary_marker)}{q2vi_summary_text}</span>
</div></pre>
""")
```

# **Objective 3: Linguistic Independence — Does the converted status depend on the preferred language?**


**Categorical Framework:** The joint distribution of preferred language and conversion outcome across all users is evaluated through a chi-square test of independence, which assesses whether conversion behavior is statistically associated with the user's preferred language.


## **Preliminary Visual Diagnostics**

```python
# ── Stripe header for visual analysis ──────────────────────────────────────────────────
hout(hban("CATEGORICAL DEPENDENCY VISUAL EXAMINATION","Conversion Rate by Preferred Language — Objective 3"))

# ──  Initializes a Single-Panel Figure for Categorical Dependency Comparison  ──────────
fig, ax = plt.subplots(figsize=(8, 4))

# ──  Computes Within-Language Conversion % and Draws Stacked Bar Chart  ────────────────
props = (pd.crosstab(df["language_preferred"], df["converted"], normalize="index") * 100)
props[["yes", "no"]].plot(
    kind="bar", stacked=True, ax=ax, color=[HPRIMARY, "#223444"],
    edgecolor="white", linewidth=1.5,
)

# ──  Canvas Design, Axis Label Customizations, and Legend Placement  ───────────────────
ax.set_title("Conversion Rate by Preferred Language", fontsize=14, fontweight="bold", pad=12, loc="center")
ax.set_xlabel("Preferred Language", fontsize=11, fontweight="bold", labelpad=12)
ax.set_ylabel("Proportion (%)", fontsize=11, fontweight="bold", labelpad=12)
ax.set_xticklabels(ax.get_xticklabels(), rotation=0, fontweight='bold')
ax.legend(["Converted (Yes)", "Not Converted (No)"], loc="upper center", bbox_to_anchor=(0.5, -0.28), ncol=2, framealpha=0.9)

# ──  Applies Uniform Annotation for Within-Bar Percentage  ─────────────────────────────
for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", label_type="center", fontweight="bold", fontsize=10, color="white")

# ──  Despine, Layout, and Render  ──────────────────────────────────────────────────────
sns.despine()
plt.tight_layout()
plt.subplots_adjust(bottom=0.27)
plt.show()
```

---

## **i — Define the Null and Alternate Hypotheses**
* **H₀:** Converted status and preferred language are independent — no association exists between conversion behavior and the user's preferred language.
* **Hₐ:** Converted status and preferred language are not independent — a real association exists between conversion behavior and the user's preferred language.
* **Direction:** Right-tailed (the chi-square statistic is non-negative by construction).

## **ii — Select the Appropriate Test**
* **Test Name:** **Chi-Square Test of Independence**
* **Rationale:** The evaluation maps the relationship between two independent categorical response variables featuring mutual exclusivity: `converted` (yes/no) and `language_preferred` (English/French/Spanish).

## **iii — Decide the Significance Level**
* $\alpha = 0.05$ (a baseline 5% maximum risk of committing a Type I error / false positive verdict is accepted).

## **iv — Collect and Prepare Data**

```python
# ── Stripe header for executive summary output ─────────────────────────────────────────
hout(hban("PARAMETRIC ASSUMPTION VERIFICATION","Contingency Table & Expected Frequency Diagnostics — Objective 3"))

# ── Contingency table and expected frequency check ─────────────────────────────────────
# Verify chi-square assumption (all expected frequencies ≥ 5)
contingency_table = pd.crosstab(df['language_preferred'], df['converted'])
r, c              = contingency_table.shape
_, _, _, expected = chi2_contingency(contingency_table)
exp_df            = pd.DataFrame(expected.round(2), index=contingency_table.index, columns=contingency_table.columns)
min_exp = expected.min()

# ──  Guardrail Verdict and Dynamic Marker / Text Strings  ──────────────────────────────
q3iv_passed = min_exp >= 5
q3iv_summary_marker = "[X]" if q3iv_passed else "[ ]"
q3iv_summary_text = " — Accepted" if q3iv_passed else " — Requirements Not Satisfied"

# ──  Renders Diagnostic Summary: Dimensions, Min Expected Frequency, and Verdict ───────
hout(f"""
<pre><div style="display: flex; justify-content: space-between; max-width: 620px; margin-bottom: 4px;">
    <span>&bull; Dimensions: {hval(r)} rows &times; {hval(c)} columns &rarr; df = {hval((r-1)*(c-1))}</span>
</div></pre>
<pre><div style="display: flex; justify-content: space-between; max-width: 620px; margin-bottom: 4px;">
    <span>&bull; Minimum expected frequency: {hval(f'{min_exp:.2f}')}</span>
</div></pre>
<hr>
<pre><div style="display: flex; justify-content: space-between; font-weight: bold; max-width: 620px; margin-bottom: 4px;">
    <span>All expected frequencies &ge; 5 (Chi-square approx valid)</span>
    <span>{hval(q3iv_summary_marker)}{q3iv_summary_text}</span>
</div></pre>
""")

# ──  Displays Observed and Expected Contingency Tables ─────────────────────────────────
display(contingency_table)
display(exp_df)
```

## **v — Calculate the test statistic and *p*-value**

```python
# ── Stripe header for hypothesis test results ──────────────────────────────────────────
hout(hban("HYPOTHESIS TEST RESULTS","Chi-Square Test of Independence Execution — Objective 3"))

# ── Chi-Square Test of Independence ────────────────────────────────────────────────────
# Tests whether conversion outcome is associated with preferred language (cat × cat)
chi2_q3, p_val_q3, dof_q3, _ = chi2_contingency(contingency_table)

hout(
    f"Executing a Chi-Square Test of Independence on the categorical contingency data yields a "
    f"formal χ²-statistic of {hval(f'{chi2_q3:.4f}')} (df={hval(dof_q3)}) and a corresponding <i>p</i>-value of {hval(f'{p_val_q3:.4g}')}."
)
```

## **vi — Compare the *p*-value with $α$**

```python
# ── Stripe header for significance evaluation ──────────────────────────────────────────
hout(hban("SIGNIFICANCE EVALUATION","Alpha Threshold Comparison — Objective 3"))

# ── Compare p-value to significance level ──────────────────────────────────────────────
# Decision rule: reject H₀ if p_val_q3 < α
decision_q3 = 'FAIL TO REJECT the Null Hypothesis' if p_val_q3 >= alpha else 'REJECT the Null Hypothesis'

hout(
    f"Comparing the resulting <i>p</i>-value of {hval(f'{p_val_q3:.4g}')} against the established significance level of "
    f"α = {hval(alpha)} results in a mathematical decision to {hval(decision_q3)}."
)
```

## **vii — Chi-Square Test of Independence Observational Inferences**

* **Hypothesis Resolution:** The chi-square test returned a *p*-value of approximately **0.213**, greater than the $α$ = 0.05 threshold. The null hypothesis of independence is not rejected.
* **Interpretation with Statistical Boundaries:** This result does *not* prove that conversion and preferred language are independent at the population level; it indicates the present sample (n = 100, with descriptive per-language conversion rates of approximately **65.6%** English, **44.1%** French, **52.9%** Spanish) does not provide sufficient evidence to reject independence at the 5% significance level. The 21.5-percentage-point descriptive gap between English and French is meaningful in business terms but is not statistically distinguishable from chance variation at this sample size — an absence of significance best interpreted as a power limitation rather than confirmed cross-linguistic uniformity. Any decision to forgo language-specific personalization should be revisited at a larger sample.

# **Objective 4: Cross-Linguistic Engagement — Is the time spent on the new page the same for different language users?**


**Group-Mean Framework:** Engagement time on the new landing page is compared across English, French, and Spanish users via a one-way ANOVA to assess whether mean time-on-page differs meaningfully by language group.


## **Preliminary Visual Diagnostics**

```python
# ── Stripe header for visual analysis ──────────────────────────────────────────────────
hout(hban("CONTINUOUS ENGAGEMENT VISUAL EXAMINATION","Time Spent Distribution by Preferred Language — Objective 4"))

# ── Subsets to New-Page Users and Initializes Two-Panel Figure ─────────────────────────
df_new = df[df['landing_page'] == 'new'].copy()
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# ── Builds Per-Language Time-on-Page Series ────────────────────────────────────────────
lang_data = [df_new[df_new['language_preferred'] == l]['time_spent_on_the_page']
             for l in ['English', 'French', 'Spanish']]

# ── Left Panel: Builds Horizontal Boxplot with Mean Marker and Custom palette detail ───
bp = axes[0].boxplot(lang_data, labels=['English', 'French', 'Spanish'],
                     patch_artist=True, vert=False, widths=0.7, showmeans=True,
                     medianprops={'color': ACCENT, 'linewidth': 2.5},
                     whiskerprops={'color': ACCENT, 'linewidth': 1.5},
                     capprops={'color': ACCENT, 'linewidth': 1.5},
                     meanprops={'marker': 'D', 'markerfacecolor': ACCENT,
                                'markeredgecolor': 'black', 'markersize': 8})

# ── Left Panel: Canvas and Text Customization ──────────────────────────────────────────
for patch, color in zip(bp['boxes'], C3PALETTE):
    patch.set_facecolor(color); patch.set_alpha(0.75)
axes[0].set_title('Time on New Page by Language', fontsize=14, fontweight='bold', pad=20)
axes[0].title.set_position([0.45, 1.05])
axes[0].set_xlabel('Time Spent (minutes)', fontsize=11, fontweight='bold', labelpad=12)
axes[0].set_ylabel('Preferred Language', fontsize=11, fontweight='bold', labelpad=12)
sns.despine(ax=axes[0])

# ── Right Panel: Computes Per-Language Mean and Builds Annotated Bar Chart ─────────────
lang_means = (df_new.groupby('language_preferred')['time_spent_on_the_page']
                    .mean().reindex(['English', 'French', 'Spanish']))
bars = axes[1].barh(lang_means.index, lang_means.values, color=C3PALETTE,
                    edgecolor='black', linewidth=1.2)
for bar, val in zip(bars, lang_means.values):
    axes[1].text(bar.get_width() / 2, bar.get_y() + bar.get_height() / 2,
                 f'{val:.2f} min', ha='center', va='center',
                 fontweight='bold', fontsize=11, color='white')

# ── Right Panel: Canvas and Text Customization ─────────────────────────────────────────
axes[1].set_title('Mean Time on New Page by Language', fontsize=14, fontweight='bold', pad=20)
axes[1].title.set_position([0.45, 1.05])
axes[1].set_xlabel('Mean Time (minutes)', fontsize=11, fontweight='bold', labelpad=12)
axes[1].set_ylabel('Preferred Language', fontsize=11, fontweight='bold', labelpad=12)
axes[1].set_xlim(0, 9)
sns.despine(ax=axes[1])
plt.tight_layout()
plt.show()
```

---

## **i — Define the Null and Alternate Hypotheses**
* **H₀:** $\mu_{\text{English}} = \mu_{\text{French}} = \mu_{\text{Spanish}}$ — Mean time on the new page is equal across all language groups.
* **Hₐ:** At least one language group has a different mean time on the new page.
* **Direction:** Right-tailed (F-statistic is always non-negative)

## **ii — Select the Appropriate Test**
* **Test Name:** **One-Way ANOVA**
* **Rationale:** One continuous response variable (`time_spent_on_the_page`) across three independent groups (English, French, Spanish), subset to new-page users only. If H₀ is rejected, a **Tukey HSD** post-hoc identifies which pairs differ.

## **iii — Decide the Significance Level**
* $\alpha = 0.05$ (a baseline 5% maximum risk of committing a Type I error / false positive verdict is accepted).

## **iv — Collect and prepare data**

```python
# ── Per-group data and ANOVA assumption checks ─────────────────────────────────────────
# Check normality (Shapiro) across three language groups and equal variance (Levene) to choose ANOVA vs Kruskal-Wallis
time_en = df_new[df_new['language_preferred'] == 'English']['time_spent_on_the_page']
time_fr = df_new[df_new['language_preferred'] == 'French']['time_spent_on_the_page']
time_es = df_new[df_new['language_preferred'] == 'Spanish']['time_spent_on_the_page']

summary_q4 = (df_new.groupby('language_preferred')['time_spent_on_the_page']
                    .agg(['count','mean','std'])
                    .reindex(['English','French','Spanish'])
                    .round(4))

w_en, p_en       = shapiro(time_en)
w_fr, p_fr       = shapiro(time_fr)
w_es, p_es       = shapiro(time_es)
lev_q4, p_lev_q4 = levene(time_en, time_fr, time_es)

# ── Dynamic verdict helpers driven by project-wide alpha ───────────────────────────────
def norm_verdict(p):
    return f'{hval("[X]")} — Normal' if p > alpha else f'{hval("[ ]")} — Non-normal'

var_q4          = f'{hval("[X]")} — Equal variances' if p_lev_q4 > alpha else f'{hval("[ ]")} — Unequal variances'
independence_ok = True
indep_verdict   = f'{hval("[X]")} — Accepted' if independence_ok else f'{hval("[ ]")} — Not Accepted'
_grand          = df_new['time_spent_on_the_page'].mean()

# ── Consolidated verdict split into right-column marker + left sub-line procedure ──────
all_normal_q4     = (p_en > alpha) and (p_fr > alpha) and (p_es > alpha)
equal_var_q4_pass = p_lev_q4 > alpha
if all_normal_q4 and equal_var_q4_pass and independence_ok:
    verdict_marker    = f'{hval("[X]")} — All ANOVA assumptions met'
    verdict_procedure = 'One-way ANOVA is the appropriate inferential procedure'
else:
    verdict_marker    = f'{hval("[ ]")} — One or more ANOVA assumptions failed'
    verdict_procedure = 'consider Kruskal-Wallis as a non-parametric alternative'

# ── Label-to-marker column so every [X] vertically stacks at the same position ─────────
L_grand  = " • Grand mean (new page users):"
L_en     = " • Shapiro-Wilk (English):"
L_fr     = " • Shapiro-Wilk (French):"
L_es     = " • Shapiro-Wilk (Spanish):"
L_lev    = " Levene's Test for Homogeneity of Variance"
L_indep  = " Independence Requirement Satisfied (Random Assignment)"
L_verd   = " Final assumption verdict"

# ── Parametric Assumption Verification logs (stripe banner + monospace diagnostics) ────
hout(
    f'<div style="max-width: 620px; border-left: 4px solid #A31D6B; padding-left: 12px; margin-bottom: 8px; margin-top: 0.8em;">'
        f'<b style="font-size: 18px; color: inherit; font-weight: bold;">PARAMETRIC ASSUMPTION VERIFICATION</b>'
        f'<br><span style="color: #666; font-size: 14px;">Sample Statistics and Normality / Variance Diagnostics — Objective 4</span>'
    f'</div>'
    f"<pre style='font-family:Menlo, Monaco, \"Courier New\", monospace; font-size:14px; line-height:1.7; color: inherit; background:transparent; padding:0; border:none; margin:0; max-width: 720px;'>"
        f"{L_grand.ljust(MARKER_COL)}{hval(f'{_grand:.4f} min')}\n"
        f"{L_en.ljust(MARKER_COL)}{norm_verdict(p_en)}\n"
        f"   W={hval(f'{w_en:.4f}')}, p={hval(f'{p_en:.4f}')}\n"
        f"{L_fr.ljust(MARKER_COL)}{norm_verdict(p_fr)}\n"
        f"   W={hval(f'{w_fr:.4f}')}, p={hval(f'{p_fr:.4f}')}\n"
        f"{L_es.ljust(MARKER_COL)}{norm_verdict(p_es)}\n"
        f"   W={hval(f'{w_es:.4f}')}, p={hval(f'{p_es:.4f}')}"
        f"<hr style='max-width: 720px; margin: 8px 0; border: 0; border-top: 1px solid #ddd;'>"
        f"<b>{L_lev.ljust(MARKER_COL)}{var_q4}</b>\n"
        f"<b>{L_indep.ljust(MARKER_COL)}{indep_verdict}</b>"
        f"<hr style='max-width: 720px; margin: 8px 0; border: 0; border-top: 1px solid #ddd;'>"
        f"<b>{L_verd.ljust(MARKER_COL)}{verdict_marker}\n"
        f" {verdict_procedure}</b>"
    f"</pre>"
)
display(summary_q4)
```

## **v — Calculate the Test Statistic and *p*-value**

```python
# ── Stripe header for hypothesis test results ──────────────────────────────────────────
hout(hban("HYPOTHESIS TEST RESULTS","One-Way ANOVA Execution — Objective 4"))

# ── One-Way ANOVA ──────────────────────────────────────────────────────────────────────
# Tests whether mean time-on-page differs across English/French/Spanish (new-page users)
f_stat_q4, p_val_q4 = f_oneway(time_en, time_fr, time_es)

hout(
    f"With the parametric assumptions documented above satisfied, the one-way ANOVA is the appropriate inferential procedure. "
    f"Executing the test on the continuous engagement data yields a formal F-statistic of {hval(f'{f_stat_q4:.4f}')} "
    f"and a corresponding <i>p</i>-value of {hval(f'{p_val_q4:.4g}')}."
)
```

## **vi — Compare the *p*-value with $α$**

```python
# ── Stripe header for significance evaluation ──────────────────────────────────────────
hout(hban("SIGNIFICANCE EVALUATION","Alpha Threshold Comparison — Objective 4"))

# ── Compare p-value to significance level ──────────────────────────────────────────────
# Decision rule: reject H₀ if p_val_q4 < α
decision_q4 = 'FAIL TO REJECT the Null Hypothesis' if p_val_q4 >= alpha else 'REJECT the Null Hypothesis'

hout(
    f"Comparing the resulting <i>p</i>-value of {hval(f'{p_val_q4:.4g}')} against the established significance level of "
    f"α = {hval(alpha)} results in a mathematical decision to {hval(decision_q4)}."
)
```

## **vii — One-Way ANOVA Observational Inferences**

* **Hypothesis Resolution:** With all parametric assumptions verified (normality across all three language subgroups and equal variances by Levene's test), the one-way ANOVA returned a *p*-value of approximately **0.432**, greater than the $α$ = 0.05 threshold. The null hypothesis of equal group means is not rejected.
* **Interpretation with Statistical Boundaries:** This result does *not* prove the three language groups spend identical time on the new page at the population level; it indicates the sample (n ≈ 17 per language group on the new page) does not provide sufficient evidence to detect mean differences at the 5% significance level. The descriptive group means observed on the new page — **English ≈ 6.66 min**, **French ≈ 6.20 min**, **Spanish ≈ 5.84 min** — span a roughly 0.8-minute range that is small in business terms but not negligible. With per-group sub-samples this small, the test has limited power to flag moderate differences; any conclusion of cross-linguistic uniformity in engagement time should be confirmed at a larger sample before localization investment decisions are scoped out.

Run a follow-up component-level isolation study to identify which elements contribute what proportion of the engagement gain, double down on the highest-leverage ones, and propagate them to other surfaces (registration, checkout, content pages).

4. **Qualitative Feedback Loop:** The four Objectives quantified *what* the new page does (engagement up, conversion up, language-neutral within power); they cannot surface *why*. Deploy short post-conversion surveys to capture user-articulated reasons for engagement, giving the product team the qualitative signal for the next design iteration that statistical tests alone cannot provide.

5. **Benchmarking and Continuous A/B Capability:** Treat the four Objective outcomes confirmed in this study — the **6.22-minute** engagement mean, the **66%** treatment conversion rate, and the language-neutral pattern within the present sample — as the operational benchmark going forward. Stand up a continuous A/B testing pipeline using the same null/alternative-framing → test-selection → *p*-value → inference workflow demonstrated here, so that every subsequent design change is validated against this benchmark before company-wide rollout. The pipeline also serves as an early-warning system: any sustained drop in engagement or conversion below the benchmark surfaces as a measurable deviation requiring investigation rather than going unnoticed.

# **APPENDIX: DEEPER MULTIVARIATE ANALYSIS**

## **Logistic Regression, Engagement Buckets, Regional Chi-Square**

The univariate, bivariate, and hypothesis testing layers describe variables in clean isolation or strict pairs. Real product experiments rarely live in such clean isolation—the variables interact.

The primary A/B test results establish the core findings. The supplementary analyses below *fortify* those findings via independent statistical lenses. To execute this, the multivariate layer derives new variables, builds a correlation heatmap, deploys predictive-modeling validation (logistic regression), and runs regional robustness checks.

None of these supplant the main hypothesis tests; rather, they triangulate the same conclusions from converging angles and surface intervention opportunities the headline tests do not reveal on their own.

## **Derived Variables for Multivariate Analysis**

The following table documents the seven columns appended to the working DataFrame across the analysis. `converted_binary`, `time_rank`, and `rank_decile` are created in the bivariate decile heatmap cell; `ratio` is created in the multivariate KDE density cell; `group_binary`, `engagement_bucket`, and `region` are introduced here in the appendix.

| **Variable** | **Data Type** | **Measurement Type** | **Description** |
| :--- | :--- | :--- | :--- |
| **`converted_binary`** | int64 | Categorical (Binary) | Numeric encoding of conversion outcome for correlation and ranking analyses<br>Field values:<br>• 0 — Visitor did not convert<br>• 1 — Visitor did convert |
| **`time_rank`** | int64 | Numerical (Ordinal) | Sequential rank 1–100 of `time_spent_on_the_page` via `rank(method='first')`. Enables the decile view and Spearman correlation |
| **`rank_decile`** | category | Categorical (Ordinal) | Buckets `time_rank` into ten ordered deciles for the heatmap<br>Field values:<br>• D1 through D10 |
| **`ratio`** | float64 | Numerical (Continuous) | Per-row engagement ratio (`time_spent_on_the_page / pop_mean`) used in the KDE comparison |
| **`group_binary`** | int64 | Categorical (Binary) | Numeric encoding of treatment assignment for the correlation matrix and logistic regression<br>Field values:<br>• 0 — Control<br>• 1 — Treatment |
| **`engagement_bucket`** | category | Categorical (Ordinal) | Three-tier view of time-on-page for the engagement × group cross-tab<br>Field values:<br>• Low (< 3 min)<br>• Medium (3–6 min)<br>• High (≥ 6 min) |
| **`region`** | object | Categorical (Nominal) | Language-derived regional cluster used in the Regional Chi-Square corroboration<br>Field values:<br>• Anglophone (U.S.)<br>• Francophone (Canada / France)<br>• Hispanophone (Latin America / Spain) |

```python
# ── Appendix derived-variables construction log ────────────────────────────────────────
hout(hban("DERIVED VARIABLE CONSTRUCTION","Binary Encodings and Engagement Bucketing for Multivariate Analysis"))

# ── Encode 'group' as a 0/1 binary (control = 0, treatment = 1) ────────────────────────
df['group_binary'] = (df['group'] == 'treatment').astype(int)

# ── Bucket the continuous time variable into three engagement tiers ────────────────────
df['engagement_bucket'] = pd.cut(
    df['time_spent_on_the_page'],
    bins=[-0.01, 3.0, 6.0, df['time_spent_on_the_page'].max() + 0.01],
    labels=['Low (<3 min)', 'Medium (3–6 min)', 'High (≥6 min)']
)

# ── Summary callout describing the new columns and previewing the DataFrame ────────────
hout(
    "Three derived variables have been added to the working DataFrame. The two binary encodings make the "
    "categorical predictors usable in the correlation heatmap below, while the engagement bucket provides "
    "a categorical view of the continuous time-on-page measure for the cross-tab that follows. The first "
    "ten rows below preview the new columns alongside the originals."
)
display(df.head(10))
```

```python
# ── Appendix correlation heatmap log ───────────────────────────────────────────────────
hout(hban("CORRELATION MATRIX HEATMAP","Pairwise Linear Relationships Across Time, Conversion, and Group"))

# ── Correlation matrix: time + two binaries (landing_page = duplicate of group) ────────
num_cols = ['time_spent_on_the_page', 'converted_binary', 'group_binary']
corr_matrix = df[num_cols].corr()
clean_labels = ['Time on Page', 'Converted (1=yes)', 'Group (1=treatment)']

# ── Render the heatmap with the established RdPu palette and gold gridlines ────────────
fig, ax = plt.subplots(figsize=(7.5, 5.5))
sns.heatmap(
    corr_matrix, annot=True, fmt='.2f', cmap='RdPu', vmin=-1, vmax=1,
    xticklabels=clean_labels, yticklabels=clean_labels,
    linecolor=ACCENT, linewidths=0.8,
    annot_kws={'fontsize': 13, 'fontweight': 'bold'}, ax=ax
)
ax.set_title('Correlation Heatmap — Time, Conversion & Treatment Assignment',
             fontsize=13, pad=14, loc='left')
plt.tight_layout(); plt.show()

# ── Pairwise correlation summary tying values to bivariate observations ────────────────
c_time_conv  = corr_matrix.loc['time_spent_on_the_page', 'converted_binary']
c_time_group = corr_matrix.loc['time_spent_on_the_page', 'group_binary']
c_conv_group = corr_matrix.loc['converted_binary', 'group_binary']
c1_s = f"{c_time_conv:.2f}"
c2_s = f"{c_time_group:.2f}"
c3_s = f"{c_conv_group:.2f}"

hout(
    f"The strongest relationship in the matrix is between time on page and conversion ({hval(c1_s)}), confirming "
    f"the bivariate observation that engaged users are more likely to convert and quantifying the strength of "
    f"that link. The treatment assignment correlates moderately with time on page ({hval(c2_s)}) and with "
    f"conversion ({hval(c3_s)}) — both in the positive direction, indicating the new page lifts engagement and "
    f"conversion in tandem rather than substituting one for the other. None of the correlations approach the "
    f"0.8+ range associated with multicollinearity concerns, so they can be treated as three real, independent "
    f"signals rather than three views of the same effect."
)
```

```python
# ── Appendix engagement × group cross-tab log ────────────────────────────────
hout(hban("ENGAGEMENT × GROUP CONVERSION CROSS-TAB","Conversion Rate by Engagement Bucket Within Control and Treatment"))

# ── Cross-tab: conversion rate by engagement bucket × group ────────────────────────────
# observed=True suppresses unused-category warnings on the categorical bucket
eng_group = (df.groupby(['engagement_bucket', 'group'], observed=True)['converted_binary']
               .agg(['count','mean'])
               .rename(columns={'count':'n','mean':'conv_rate'}))
eng_group['conv_rate'] = (eng_group['conv_rate'] * 100).round(1)
eng_group['n'] = eng_group['n'].astype(int)
display(eng_group)

# ── Visualize the same data as a grouped bar chart ─────────────────────────────────────
plot_df = eng_group.reset_index()
fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(
    data=plot_df, x='engagement_bucket', y='conv_rate', hue='group',
    order=['Low (<3 min)','Medium (3–6 min)','High (≥6 min)'],
    hue_order=['control','treatment'], palette=C2PALETTE, ax=ax,
    edgecolor='#141B24', linewidth=1.2
)

# ── Annotate each bar with its conversion rate percentage ──────────────────────────────
for p in ax.patches:
    h = p.get_height()
    if h > 0:
        ax.annotate(
            f'{h:.1f}%',
            (p.get_x() + p.get_width()/2., h + 1.5),
            ha='center', va='bottom', fontsize=10, fontweight='bold'
        )
ax.set_title('Conversion Rate by Engagement Bucket and Experiment Group',
             fontsize=13, pad=14, loc='left')
ax.set_xlabel('Engagement Bucket'); ax.set_ylabel('Conversion Rate (%)')
ax.set_ylim(0, 110)
ax.legend(title='Group', frameon=False)
plt.tight_layout(); plt.show()

# ── Two-pattern interpretation tying engagement and group effects together ─────────────
hout(
    "Two patterns stand out in the engagement × group cross-tab. First, conversion rises sharply with engagement "
    "in both groups: low-engagement users (under three minutes) convert at roughly 7 percent, mid-engagement "
    "users convert at about 35–46 percent, and high-engagement users (six or more minutes) convert at 81–85 "
    "percent. Engagement is by far the strongest predictor of conversion in the dataset. Second, within each "
    "engagement bucket, the treatment group converts at the same rate as or higher than control, indicating "
    "the new page's benefit is not concentrated in one part of the engagement distribution. The cell counts "
    "(n) shown alongside each rate are small, so these within-bucket comparisons are descriptive rather than "
    "statistically definitive at this sample size."
)
```

```python
# ── Appendix logistic regression: time + treatment → conversion ───────────────────────
hout(hban("LOGISTIC REGRESSION — TIME AND TREATMENT ON CONVERSION","Two-Predictor Probabilistic Modeling"))

# ── Fit the model: time_spent + group_binary as predictors ─────────────────────────────
X = sm.add_constant(df[['time_spent_on_the_page', 'group_binary']])
y = df['converted_binary']
logit_model = sm.Logit(y, X).fit(disp=0)

# ── Coefficient table: coef, std_err, odds ratio, p-value ──────────────────────────────
coef_df = pd.DataFrame({
    'coef':       logit_model.params.round(4),
    'std_err':    logit_model.bse.round(4),
    'odds_ratio': np.exp(logit_model.params).round(4),
    'p_value':    logit_model.pvalues.round(4),
})
display(coef_df)

# ── Extract values for narrative ───────────────────────────────────────────────────────
or_time  = np.exp(logit_model.params['time_spent_on_the_page'])
or_group = np.exp(logit_model.params['group_binary'])
p_time   = logit_model.pvalues['time_spent_on_the_page']
p_group  = logit_model.pvalues['group_binary']
pct_time = (or_time - 1) * 100

# ── Stakeholder interpretation: time is the lever, treatment is mediated ───────────────
hout(
    f"Time on page is a strong and highly significant predictor of conversion — each "
    f"additional minute multiplies the odds of converting by {hval(f'{or_time:.2f}')} "
    f"(a {hval(f'{pct_time:.0f}%')} increase per minute, <i>p</i> = {hval(f'{p_time:.4f}')}). "
    f"Once time on page is held constant, the residual effect of treatment-group "
    f"assignment yields an odds ratio of {hval(f'{or_group:.2f}')} with <i>p</i> = "
    f"{hval(f'{p_group:.4f}')}, which is not statistically significant. The implication "
    f"is that the new landing page's conversion lift is largely <b>mediated by</b> the "
    f"increased engagement time it produces, rather than acting as a direct conversion "
    f"driver independent of how long users spend on the page. Interventions that further "
    f"extend on-page engagement are therefore likely higher-leverage than additional "
    f"design swaps."
)
```

```python
# ── Appendix regional chi-square: corroborate Objective 3 under a regional framing ────
hout(hban("REGIONAL CHI-SQUARE — LANGUAGE → REGION PROXY","Objective 3 Corroboration Under an Alternate Variable Framing"))

# ── Proxy mapping: language → broad regional cluster ───────────────────────────────────
# The dataset lacks explicit geolocation, so language is used as a directional
# proxy for region. The mapping is approximated, not asserted as geography.
region_map = {
    'English': 'Anglophone (U.S.)',
    'French':  'Francophone (Canada / France)',
    'Spanish': 'Hispanophone (Latin America / Spain)',
}
df['region'] = df['language_preferred'].map(region_map)

# ── Build the region × converted contingency table ─────────────────────────────────────
region_contingency = pd.crosstab(df['region'], df['converted'])
display(region_contingency)

# ── Run the chi-square test on the regional framing ────────────────────────────────────
chi2_r, p_r, dof_r, expected_r = chi2_contingency(region_contingency)
min_exp_r = expected_r.min()

# ── Build the result summary table ─────────────────────────────────────────────────────
result_df = pd.DataFrame({
    'statistic':         [round(chi2_r, 4)],
    'degrees_of_freedom':[dof_r],
    'p_value':           [round(p_r, 4)],
    'min_expected_freq': [round(min_exp_r, 2)],
}, index=['Regional Chi-Square'])
display(result_df)

# ── Stakeholder interpretation: proxy framing + Objective 3 corroboration ─────────────
hout(
    f"Under a language-to-region proxy mapping, the chi-square test returns the same result "
    f"as Objective 3: χ² = {hval(f'{chi2_r:.4f}')}, df = {hval(dof_r)}, p = {hval(f'{p_r:.4f}')}. "
    f"This is by design — the proxy is a relabeling of the same underlying variable, so the "
    f"statistical conclusion is identical and corroborates Objective 3 under an explicit regional framing. "
    f"What the regional framing adds is interpretive: descriptively, the Anglophone (U.S.) "
    f"cluster shows the highest conversion rate while the Francophone (Canada / France) cluster "
    f"shows the lowest. Whether that reflects a real regional effect or sampling noise at "
    f"~{hval('33')} users per region cannot be adjudicated at this sample size. The proxy mapping "
    f"itself is a limitation — the dataset lacks explicit geolocation, and Spanish speakers in "
    f"particular could be in Mexico, broader Latin America, or Spain. Any regionally targeted "
    f"intervention should be confirmed against an enrichment of the dataset with actual "
    f"geolocation before resources are committed."
)
```
